```python
import numpy as np
import random



# PARAMETERS

POP_SIZE = 20
GENERATIONS = 50
CROSSOVER_RATE = 0.8
MUTATION_RATE = 0.1
R = 1000

# VARIABLE BOUNDS

X1_MIN, X1_MAX = 0, 200
X2_MIN, X2_MAX = 0, 200
X3_MIN, X3_MAX = 0, 30

# OBJECTIVE FUNCTION

def objective(x):

    x1, x2, x3 = x

    return 350*x1 + 120*x2 + 500*x3

# CONSTRAINTS

def constraints(x):

    x1, x2, x3 = x

    g1 = 0.7*x1 + 2*x3 - 80

    g2 = x2 - 30

    g3 = 0.15*x2 - x3

    g4 = 60000 - (350*x1 + 120*x2 + 500*x3)

    return [g1, g2, g3, g4]

# TOTAL CONSTRAINT VIOLATION

def total_violation(x):

    g = constraints(x)

    violation = sum(max(0, -gi) for gi in g)

    return violation

# PENALIZED FITNESS

def penalized_objective(x):

    f = objective(x)

    g = constraints(x)

    penalty = R * sum(max(0, -gi)**2 for gi in g)

    return f + penalty

# FEASIBILITY CHECK

def is_feasible(x):

    return all(g >= 0 for g in constraints(x))

# DEB'S FEASIBILITY RULES

def better(x1, x2):

    feasible1 = is_feasible(x1)
    feasible2 = is_feasible(x2)

    # Rule 1
    if feasible1 and not feasible2:
        return x1

    if feasible2 and not feasible1:
        return x2

    # Rule 2
    if feasible1 and feasible2:

        if objective(x1) < objective(x2):
            return x1
        else:
            return x2

    # Rule 3

    if total_violation(x1) < total_violation(x2):
        return x1
    else:
        return x2

# INITIAL POPULATION

def create_population():

    population = []

    for _ in range(POP_SIZE):

        x1 = random.uniform(X1_MIN, X1_MAX)
        x2 = random.uniform(X2_MIN, X2_MAX)
        x3 = random.uniform(X3_MIN, X3_MAX)

        population.append([x1, x2, x3])

    return population

# TOURNAMENT SELECTION

def selection(population):

    a = random.choice(population)
    b = random.choice(population)

    return better(a, b)

# CROSSOVER

def crossover(parent1, parent2):

    if random.random() < CROSSOVER_RATE:

        alpha = random.random()

        child1 = [
            alpha*parent1[i] + (1-alpha)*parent2[i]
            for i in range(3)
        ]

        child2 = [
            alpha*parent2[i] + (1-alpha)*parent1[i]
            for i in range(3)
        ]

        return child1, child2

    return parent1[:], parent2[:]

# MUTATION

def mutate(child):

    if random.random() < MUTATION_RATE:

        index = random.randint(0, 2)

        if index == 0:
            child[0] += random.uniform(-10, 10)

        elif index == 1:
            child[1] += random.uniform(-10, 10)

        else:
            child[2] += random.uniform(-2, 2)

    # Bound control

    child[0] = np.clip(child[0], X1_MIN, X1_MAX)
    child[1] = np.clip(child[1], X2_MIN, X2_MAX)
    child[2] = np.clip(child[2], X3_MIN, X3_MAX)

    return child

# MAIN GA

population = create_population()

best_solution = None

for generation in range(GENERATIONS):

    new_population = []

    while len(new_population) < POP_SIZE:

        parent1 = selection(population)
        parent2 = selection(population)

        child1, child2 = crossover(parent1, parent2)

        child1 = mutate(child1)
        child2 = mutate(child2)

        new_population.append(child1)
        new_population.append(child2)

    population = new_population[:POP_SIZE]

    generation_best = population[0]

    for candidate in population:

        generation_best = better(
            generation_best,
            candidate
        )

    best_solution = generation_best

    print(
        f"Generation {generation+1:02d} | "
        f"Best Cost = {objective(best_solution):.2f} | "
        f"Feasible = {is_feasible(best_solution)}"
    )

# FINAL RESULTS

x1, x2, x3 = best_solution

print("\n")
print("="*50)
print("FINAL OPTIMUM SOLUTION")
print("="*50)

print(f"x1 = {x1:.4f}")
print(f"x2 = {x2:.4f}")
print(f"x3 = {x3:.4f}")

print("\nObjective Cost:")
print(objective(best_solution))

print("\nConstraints:")

g = constraints(best_solution)

for i, value in enumerate(g, start=1):

    print(f"g{i} = {value:.4f}")

print("\nFeasible:", is_feasible(best_solution))
```
